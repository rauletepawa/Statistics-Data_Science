### Ejercicio entrenamiento 1 ###
Dada la siguiente función que calcula la suma de cuadrados, escribe el código necesario para realizar pruebas de código usando doctest.

La suma de cuadrados se refiere a dado un numero n sumar todos los cuadrados hasta llegar a él, es decir, si n=3 la suma de cuadros sera $1^{2} + 2^{2} + 3^{2}$, $1+4+9$, =14`




```python
def sum_of_squares(n):
    """Calcula la suma de los cuadrados hasta n.

        Respuesta

    """

    if not n >= 0:
        raise ValueError("n debe ser >= 0")

    result = 0
    for i in range(n + 1):
        result += i ** 2

    return result

if __name__ == "__main__":
    import doctest
    doctest.run_docstring_examples(sum_of_squares(3), globals(), verbose=True)
```

La función debe seguir los pasos adecuados para encontrar el complemento inverso de la secuencia de ADN, teniendo en cuenta las bases complementarias (A-T, C-G) y el orden inverso de la secuencia.

Referencia:

    - Complementary DNA Strands:
    
    https://www.wikihow.life/Find-the-Reverse-Complement-of-a-DNA-Sequence#:~:text=Trace%20through%20the%20sequence%20backwards,last%20nucleotide%20in%20the%20sequence.&text=As%20you%20pass%20over%20each,hand%20side%20of%20the%20page.
    

```python
# Importar librerias requeridas
import unittest
import sys

def complementaria_inversa(dna_seq):
    """
    Encuentra la complementaria inversa de una secuencia DNA.

    Parametros:
    - dna_secuencia (str): Entrada DNA.

    Return:
    - str: Complementaria inversa del DNA.

    Referencia:
    - Complementary DNA Strands:https://www.wikihow.life/Find-the-Reverse-Complement-of-a-DNA-Sequence#:~:text=Trace%20through%20the%20sequence%20backwards,last%20nucleotide%20in%20the%20sequence.&text=As%20you%20pass%20over%20each,hand%20side%20of%20the%20page.
    """
    complement = {"A": "T", "T": "A", "C": "G", "G": "C"}
    reverse_dna = dna_seq[::-1]
    complementaria_inversa = "".join(complement[base] for base in reverse_dna)
    return complementaria_inversa


#Tests
class TestEjercicio2(unittest.TestCase):
    def test_reverse_complement(self):
        self.assertEqual(complementaria_inversa("ATCG"), "CGAT")
        self.assertEqual(complementaria_inversa("AATTCCGG"), "CCGGAATT")
        self.assertEqual(complementaria_inversa(""), "")
        self.assertEqual(complementaria_inversa("A"), "T")
        self.assertEqual(complementaria_inversa("ACGT"), "ACGT")
        self.assertEqual(complementaria_inversa("AGCTTAGC"), "GCTAAGCT")
        self.assertEqual(complementaria_inversa("GGCC"), "GGCC")

    def test_invalid_characters(self):
        with self.assertRaises(KeyError):
            complementaria_inversa("ATXG")
        with self.assertRaises(KeyError):
            complementaria_inversa("123")
        with self.assertRaises(KeyError):
            complementaria_inversa("ATCGU")


if __name__ == '__main__':
    suite = unittest.TestLoader().loadTestsFromTestCase( TestEjercicio2 )
    unittest.TextTestRunner(verbosity=1,stream=sys.stderr).run( suite )
```

### Ejercicio entrenamiento 3
**Hablamos de tests en verde cuando todos nuestros tests se ejecutan correctamente y dan el resultado esperado y tests en rojo en caso contrario.**

En el siguiente ejercicio, escribe el código necesario que haga cumplir todos los tests, es decir, que los tests estén *en verde* **sin modificar en ningún momento la parte correspondiente al código de los tests** (`class TestMonster`).

```python
import unittest
import sys


class Monster:
    """A cute monster"""

    def __init__(self, name, size, basic_type='fire'):
        """Creates a new monster"""
        self.name = name
        self.size = size
        self.basic_type = basic_type.lower()

    def fight(self, monster2):
        """Two monsters fight each other. Returns the name of the monster winning
        or 'draw' otherwise.

        Monster battles have very specific rules:
        - Water wins over Fire
        - Electrical wins over Water
        - Plant wins over Electrical
        - Fire wins over Plant

        if both monsters fighting have the same type, bigger monster wins!
        if both monsters fighting have the same type and size, it's a draw

        Any other combination, it's a draw
        """
        #Respuesta

        rules = {
            'water': 'fire',
            'fire': 'plant',
            'plant': 'electrical',
            'electrical': 'water'
        }

        if self.basic_type == monster2.basic_type:
            if self.size > monster2.size:
                return self.name
            elif self.size < monster2.size:
                return monster2.name
            else:
                return 'draw'
        elif rules.get(self.basic_type) == monster2.basic_type:
            return self.name
        elif rules.get(monster2.basic_type) == self.basic_type:
            return monster2.name
        else:
            return 'draw'

class TestMonster(unittest.TestCase):

    def test_create_a_new_monster(self):
        # Create a monster
        pika = Monster("Pika", 10, "electrical")

        # Test if it is not none
        self.assertIsNotNone(pika)

        # Test if attributes have been set properly
        self.assertEqual("Pika", pika.name)
        self.assertEqual(10, pika.size)
        self.assertEqual("electrical", pika.basic_type)

    def test_fight_water_fire(self):
        """Test which monsters wins in a water vs fire fight"""
        squirtle = Monster("Squirtle", 10, "water")
        char = Monster("Char", 10, "fire")

        self.assertEqual("Squirtle", squirtle.fight(char))

    def test_fight_fire_plant(self):
        """Test which monsters wins in a plant vs fire fight"""
        char = Monster("Char", 10, "fire")
        bulb = Monster("Bulb", 15, "plant")

        self.assertEqual("Char", char.fight(bulb))

    def test_electrical_electrical_same_size(self):
        """Test which monsters wins if both have same type and size"""
        pika = Monster("Pika", 10, "electrical")
        pichu = Monster("Pichu", 10, "electrical")

        self.assertEqual("draw", pika.fight(pichu))

    def test_electrical_electrical_different_size(self):
        """Test which monsters wins if both have same type but different size"""
        pika = Monster("Pika", 10, "electrical")
        pichu = Monster("Pichu", 9, "electrical")

        self.assertEqual("Pika", pika.fight(pichu))


if __name__ == '__main__':
    suite = unittest.TestLoader().loadTestsFromTestCase(TestMonster)
    unittest.TextTestRunner(verbosity=1,stream=sys.stderr).run(suite)
```

Explica en qué consiste la Programacion orientada a objetos (POO) y cual es la diferencia entre clases e instancias. No olvides citar tus fuentes:

La Programación Orientada a Objetos (POO) es un paradigma de programación que organiza el software en objetos, que son instancias de clases. Este enfoque permite modelar entidades del mundo real de una manera más intuitiva y estructurada. En la POO la información y los datos se guardan en forma de atributos y las funciones en forma de métodos.

La POO se basa en varias características, 3 de las más importantes son:

    1. Encapsulamiento: Agrupa datos y métodos en una sola unidad llamada clase. Esto mejora la modularidad y la eficiencia del código.
    2. Herencia: Permite crear nuevas clases a partir de clases ya existentes. La clase derivada hedera los atributos de la clase padre. Facilita la  reutilización del código y la creación de jerarquías
    3. Polimorfismo: Comportamientos diferentes, asociados a objetos distintos, pueden compartir el mismo nombre; al llamarlos por ese nombre se utilizará el comportamiento correspondiente al objeto que se esté usando

Una clase es una plantilla que define atributos y métodos comunes para crear objetos, mientras que una instancia es un objeto concreto creado a partir de esa clase. La clase actúa como un plano abstracto, describiendo cómo deberían ser los objetos, mientras que la instancia es una manifestación real con valores específicos para esos atributos. Las clases proporcionan comportamientos comunes, mientras que las instancias tienen datos individuales y específicos.

¿Como se hereda una clase a partir de otra clase en Python?

Explicalo dando un ejemplo con clases llamadas "animal" y "perro"

La herencia permite a una clase (subclase) heredar atributos y métodos de otra clase (superclase), facilitando la reutilización y extensión del código. Por ejemplo, la clase Animal podría definir atributos generales como nombre y métodos como hacer_sonido(), aunque este último podría ser abstracto (pass) y ser implementado específicamente por subclases como Perro.

```python
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre

    def hacer_sonido(self):
        pass

class Perro(Animal):
    def __init__(self, nombre, raza):
        super().__init__(nombre)
        self.raza = raza

    def hacer_sonido(self):
        return "¡Guau!"
```

En este ejemplo, Perro hereda de Animal, obteniendo el atributo nombre y la capacidad de definir su propio sonido (hacer_sonido()). La superclase Animal puede tener otras subclases como Gato, cada una con sus propias características únicas.

```python
import unittest
import sys

def even_or_odd(int_number):
    try:
        if type(int_number) is not int:
            raise ValueError("The input is not an integer number!!!")
        if int_number % 2 == 0:
            return "The number is even"
        else:
            return "The number is odd"
    except:
        raise ValueError("The input is not an integer number!!!")


class TestEvenOrOdd(unittest.TestCase):

    def test_odd(self):
        result = even_or_odd(3)
        self.assertEqual(result, "The number is odd")

    def test_even(self):
        result = even_or_odd(2)
        self.assertEqual(result, "The number is even")

    def test_non_integer(self):
        with self.assertRaises(ValueError):
            even_or_odd("string")

    def test_zero(self):
        result = even_or_odd(0)
        self.assertEqual(result, "The number is even")

if __name__ == '__main__':
    suite = unittest.TestLoader().loadTestsFromTestCase(TestEvenOrOdd)
    unittest.TextTestRunner(verbosity=1, stream=sys.stderr).run(suite)
```

```python
import unittest
import sys

class MathOperations:
    """Clase que realiza operaciones matemáticas básicas."""

    def __init__(self, num1, num2):
        """Inicializa el objeto con dos números."""
        self.num1 = num1
        self.num2 = num2

    def add(self):
        """Retorna la suma de los dos números."""
        return self.num1 + self.num2

    def subtract(self):
        """Retorna la resta del segundo número al primero."""
        return self.num1 - self.num2

    def multiply(self):
        """Retorna el producto de los dos números."""
        return self.num1 * self.num2

    def divide(self):
        """Retorna la división del primer número entre el segundo."""
        if self.num2 == 0:
            raise ValueError("No se puede dividir por cero")
        return self.num1 / self.num2

class TestMathOperations(unittest.TestCase):

    def test_create_object(self):
        math_obj = MathOperations(5, 3)
        self.assertIsNotNone(math_obj)

    def test_addition(self):
        math_obj = MathOperations(5, 3)
        self.assertEqual(math_obj.add(), 8)

    def test_subtraction(self):
        math_obj = MathOperations(5, 3)
        self.assertEqual(math_obj.subtract(), 2)

    def test_multiplication(self):
        math_obj = MathOperations(5, 3)
        self.assertEqual(math_obj.multiply(), 15)

    def test_division(self):
        math_obj = MathOperations(6, 2)
        self.assertEqual(math_obj.divide(), 3)

    def test_division_by_zero(self):
        math_obj = MathOperations(5, 0)
        with self.assertRaises(ValueError):
            math_obj.divide()


if __name__ == '__main__':
    suite = unittest.TestLoader().loadTestsFromTestCase(TestMathOperations)
    unittest.TextTestRunner(verbosity=1, stream=sys.stderr).run(suite)
```

Dada la siguiente función que calcula el factorial de un número entero, escribe el código necesario para realizar pruebas de código usando `doctest`:



```python
import math

def factorial(n):
    """Devuelve el factorial de n, un numero entero >= 0

    Ejemplos:
    
    >>> factorial(0)
    1

    >>> factorial(5)
    120

    >>> factorial(10)
    3628800

    >>> factorial(-1)
    Traceback (most recent call last):
        ...
    ValueError: n debe ser >= 0

    >>> factorial(3.5)
    Traceback (most recent call last):
        ...
    ValueError: n tiene que ser un numero entero

    >>> factorial(1e50)
    Traceback (most recent call last):
        ...
    OverflowError: n es demasiado grande

    """

    if not n >= 0:
        raise ValueError("n debe ser >= 0")

    if math.floor(n) != n:
        raise ValueError("n tiene que ser un numero entero")

    if n+1 == n:  # capturar un valor como 1e50, 1e100,etc
        raise OverflowError("n es demasiado grande")

    result = 1
    factor = 2
    while factor <= n:
        result *= factor
        factor += 1
    return result


if __name__ == "__main__":
    import doctest
    doctest.run_docstring_examples(factorial, globals(), verbose=True)
```

```python
import unittest
import sys


class Quaternion:

    def __init__(self, w=1., x=0., y=0., z=0.):
        """Crea un cuaternión. Devuelve por defecto el cuaternión unitario."""
        self.w = w
        self.x = x
        self.y = y
        self.z = z

    def __neg__(self):
        """Negación de un cuaternión.

        Se llama: q2 = -q1
        """
        return Quaternion(-self.w, -self.x, -self.y, -self.z)

    def __add__(self, other):
        """Implementa la suma de cuaterniones.

        Ejemplo:
        q1 = Quaternion()
        q2 = Quaternion()

        q3 = q1 + q2

        """
        return Quaternion(self.w + other.w, self.x + other.x, self.y + other.y, self.z + other.z)

    def __sub__(self, other):
        """Implementa la resta de cuaterniones."""
        return Quaternion(self.w - other.w, self.x - other.x, self.y - other.y, self.z - other.z)

    def __mul__(self, scalar):
        """Implementa la multiplicación por un escalar."""
        return Quaternion(self.w * scalar, self.x * scalar, self.y * scalar, self.z * scalar)

    def __eq__(self, other):
        """Implementa la igualdad entre cuaterniones."""
        return (self.w == other.w) and (self.x == other.x) and (self.y == other.y) and (self.z == other.z)


class TestQuaternion(unittest.TestCase):
    """Tests for the Quaternion class."""

    def test_default_quaternion_creation(self):
        q = Quaternion()
        self.assertEqual(q.w, 1.0)
        self.assertEqual(q.x, 0.0)
        self.assertEqual(q.y, 0.0)
        self.assertEqual(q.z, 0.0)

    def test_quaternion_creation(self):
        q = Quaternion(0.1, 0.5, -0.5, 1.0)
        self.assertEqual(q.w, 0.1)
        self.assertEqual(q.x, 0.5)
        self.assertEqual(q.y, -0.5)
        self.assertEqual(q.z, 1.0)

    def test_negation(self):
        q = Quaternion(0.1, 0.5, -0.5, 1.0)
        neg_q = -q
        self.assertEqual(neg_q.w, -0.1)
        self.assertEqual(neg_q.x, -0.5)
        self.assertEqual(neg_q.y, 0.5)
        self.assertEqual(neg_q.z, -1.0)

    def test_addition(self):
        q1 = Quaternion(1, 2, 3, 4)
        q2 = Quaternion(0.5, -1, 2, 0)
        q3 = q1 + q2
        self.assertEqual(q3.w, 1.5)
        self.assertEqual(q3.x, 1)
        self.assertEqual(q3.y, 5)
        self.assertEqual(q3.z, 4)

    def test_subtraction(self):
        q1 = Quaternion(1, 2, 3, 4)
        q2 = Quaternion(0.5, -1, 2, 0)
        q3 = q1 - q2
        self.assertEqual(q3.w, 0.5)
        self.assertEqual(q3.x, 3)
        self.assertEqual(q3.y, 1)
        self.assertEqual(q3.z, 4)

    def test_multiplication_by_scalar(self):
        q = Quaternion(1, 2, 3, 4)
        q_scaled = q * 2
        self.assertEqual(q_scaled.w, 2)
        self.assertEqual(q_scaled.x, 4)
        self.assertEqual(q_scaled.y, 6)
        self.assertEqual(q_scaled.z, 8)

    def test_equality(self):
        q1 = Quaternion(1, 2, 3, 4)
        q2 = Quaternion(1, 2, 3, 4)
        q3 = Quaternion(0.5, -1, 2, 0)
        self.assertTrue(q1 == q2)
        self.assertFalse(q1 == q3)


if __name__ == '__main__':
    suite = unittest.TestLoader().loadTestsFromTestCase(TestQuaternion)
    unittest.TextTestRunner(verbosity=1, stream=sys.stderr).run(suite)
```

```python
import unittest
import sys

import unittest

class Peptide:
    valid_amino_acids = set("ARNDCEQGHILKMFPSTWY")

    def __init__(self, sequence):
        """Inicializa un péptido con una secuencia de aminoácidos válida."""
        if not all(aa in self.valid_amino_acids for aa in sequence):
            raise ValueError("La secuencia contiene aminoácidos no válidos.")
        self.sequence = sequence

    def concatenate(self, other):
        """Concatena dos péptidos y devuelve un nuevo péptido."""
        if not isinstance(other, Peptide):
            raise TypeError("Se espera un objeto de tipo Peptide.")
        concatenated_sequence = self.sequence + other.sequence
        return Peptide(concatenated_sequence)

    def polyh(self, num_histidines):
        """Añade un número dado de histidinas al final del péptido."""
        if not isinstance(num_histidines, int) or num_histidines < 0:
            raise ValueError("El número de histidinas debe ser un entero no negativo.")
        histidine_tail = 'H' * num_histidines
        extended_sequence = self.sequence + histidine_tail
        return Peptide(extended_sequence)

    def __eq__(self, other):
        """Compara dos péptidos por su secuencia de aminoácidos."""
        if not isinstance(other, Peptide):
            return False
        return self.sequence == other.sequence

class TestPeptide(unittest.TestCase):

    def test_valid_peptide_creation(self):
        # Verificar inicialización con una cadena válida
        peptide = Peptide("PPSSTY")
        self.assertEqual(peptide.sequence, "PPSSTY")

    def test_invalid_peptide_creation(self):
        # Verificar error al inicializar con una cadena no válida
        with self.assertRaises(ValueError):
            Peptide("PPXDNR")

    def test_concatenate(self):
        # Verificar la concatenación de dos péptidos
        peptide1 = Peptide("ARARAR")
        peptide2 = Peptide("TGGG")
        result = peptide1.concatenate(peptide2)
        self.assertEqual(result.sequence, "ARARARTGGG")

    def test_polyh(self):
        # Verificar la adición de histidinas al final
        peptide = Peptide("ARTART")
        result = peptide.polyh(5)
        self.assertEqual(result.sequence, "ARTARTHHHHH")

    def test_polyh_zero_histidines(self):
        # Verificar que polyh con 0 histidinas no modifique la secuencia
        peptide = Peptide("ARTART")
        result = peptide.polyh(0)
        self.assertEqual(result.sequence, "ARTART")

    def test_polyh_large_number_histidines(self):
        # Verificar que polyh maneje correctamente un gran número de histidinas
        peptide = Peptide("ARTART")
        result = peptide.polyh(100)
        self.assertEqual(result.sequence[-100:], "H" * 100)

    def test_equality(self):
        # Verificar la igualdad entre péptidos
        peptide1 = Peptide("PPSSTY")
        peptide2 = Peptide("PPSSTY")
        peptide3 = Peptide("ARARAR")
        self.assertEqual(peptide1, peptide2)
        self.assertNotEqual(peptide1, peptide3)

if __name__ == '__main__':
    suite = unittest.TestLoader().loadTestsFromTestCase(TestPeptide)
    unittest.TextTestRunner(verbosity=1, stream=sys.stderr).run(suite)
```

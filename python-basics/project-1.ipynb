## Tests

Abans de res, dir que en l'àmbit de la programació és molt comú utilitzar tests (anomenats tests unitaris) durant la codificació dels programes. Els tests serveixen per a saber si un programa es comporta com s'espera. És a dir, serveixen per a validar la funcionalitat del programa, no per a determinar la qualitat del codi.

Així doncs, els tests són un gran aliat del programador. És per això que la majoria dels exercicis i fins i tot alguna de les preguntes tindran una cel·la sota elles amb un comentari en la primera línia amb el títol: `# Test Exercici X` o `Test Pregunta X`

* Si el test falla, llavors el vostre codi no funciona com s'espera. Així doncs, haureu de corregir-ho.

* Si el test no falla, llavors el vostre codi es comporta com s'espera. Tanmateix, això no significa que l'el exercici sigui correcte, ja que hi ha situacions en les quals es demana usar una determinada funció o estructura. Per exemple, imagineu que es demana resoldre un exercici usant un ``for`` i, en el seu lloc, useu un ``while``. Segurament el vostre programa passaria el test (perquè funciona com s'espera), però no respecta un requisit de l'enunciat.

### Exercici 4 **<font color="green" size="+2">MU</font>**


Escriu un programa on es defineixi un **diccionari** (estructura de dades Python) anomenat `animals` les **claus dels quals** siguin *tots** els noms dels continents i el *valor** una *llista** (estructura de dades Python) amb animals endèmics d'aquest continent. Per exemple:

```
animals -> Europa: Linx ibèric, Bison Europeu, Vaca Suïssa
           Àfrica: León, Elefant Africà, Estruç
           Àsia: Pandes Gegants, Panda vermell, Tigre de bengala
           Amèrica del Nord: Aguiles calbes, Alcis americans, Bisons americans
           Amèrica del Sud: Puma, Tapir, Boa
           Antàrtida: Pingüins Rei, Foca antàrtica, Kril antàrtic
           Oceania: Kiwi comú, Dimoni de Tasmània, Coala
```

Un diccionari i una llista són estructures de dades de Python, si us plau repassa la teoria si no el recordes.

````python
animals = {'Europa': ['Linx ibèric', 'Bison Europeu', 'Vaca Suïssa'],
           'Àfrica': ['León','Elefant Africà', 'Estruç'],
           'Àsia': ['Pandes Gegants', 'Panda vermell', 'Tigre de bengala'],
           'Amèrica del Nord': ['Aguiles calbes', 'Alcis americans', 'Bisons americans'],
           'Amèrica del Sud': ['Puma', 'Tapir', 'Boa'],
           'Antàrtida': ['Pingüins Rei', 'Foca antàrtica', 'Kril antàrtic'],
           'Oceania': ['Kiwi comú', 'Dimoni de Tasmània', 'Coala']}
````

**Test**: Utilitza el test que hi ha a continuació per a comprovar que el teu programa es comporta com s'espera.

````python
%%ipytest
# Test exercici 4
def test_longitud_diccionari():
  assert len(animals.keys()) >= 5

def test_keys():
  assert set(animals.keys()).issubset({'Europa', 'Àfrica', 'Àsia', 'Amèrica del Nord', 'Amèrica del Sud', 'Antàrtida', 'Oceania'})

def test_values():
  assert all(isinstance(value, list) and len(value) >=1 for value in animals.values())
  assert all(isinstance(item, str) for value in animals.values() for item in value)
````
